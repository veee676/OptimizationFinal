### data downloand and stitching
##### comment out after first run

```python
%pip install requests
import io
import time
import requests
import pandas as pd

BASE = "https://health.data.ny.gov/resource/82xm-y6g8.csv"
LIMIT = 50000          # rows per request
OUT = "sparcs.csv"

offset = 0
first = True

while True:
    params = {"$limit": LIMIT, "$offset": offset, "$order": ":id"}  # stable ordering so pages don't overlap
    r = requests.get(BASE, params=params, timeout=180)
    r.raise_for_status()

    df = pd.read_csv(io.StringIO(r.text), dtype=str)  # dtype=str keeps codes like "0012" intact
    if df.empty:
        break

    df.to_csv(OUT, mode="w" if first else "a", header=first, index=False)
    first = False
    offset += len(df)
    print(f"Downloaded {offset:,} rows")

    if len(df) < LIMIT:
        break
    time.sleep(0.5)  # be polite to the server
```
